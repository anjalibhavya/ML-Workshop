```python
import plotly as py
import plotly.graph_objects as go
from plotly.offline import init_notebook_mode,iplot
init_notebook_mode(connected=True)
```

```python
#let's visualize histogram
trace=go.Histogram(x=data['Age'],nbinsx=20,marker={'color':'red'})
#layout define for graph
layout=go.Layout(title="Age Distribution")
#we combine the trace with layout to display the  figure
fig=go.Figure(data=trace,layout = layout)
fig.show()
```

```python
#Now we will display for Piechart
trace=go.Pie(labels=['Dead','Alive'],values=[549,342])
layout=go.Layout(title='Survived Distribution')
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#Box plot and scatter plot
trace=go.Box(x=data['Pclass'],y=data['Age'],marker={'color':'red'})
layout=go.Layout(title="Age vs Pclass Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#scatter plot representation -->Age vs Fare
trace=go.Scatter(x=data['Age'],y=data['Fare'],mode='markers',marker={'color':'black'})
layout=go.Layout(title="Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```
